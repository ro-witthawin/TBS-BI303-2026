## 1. What Is a String?

A **string** is text data in Python. You can create a string by putting text inside single quotes (`'...'`) or double quotes (`"..."`).

````python
# Creating strings
course_name = "Business Intelligence"
student_name = 'Witthawin'

print(course_name)
print(student_name)
print(type(course_name))
````

````python
# More string examples
city = "Bangkok"
country = "Thailand"
empty_text = ""

print(city)
print(country)
print(empty_text)
print(type(empty_text))
````

## 2. String Concatenation

You can join strings together using the `+` operator. This is called **concatenation**.

````python
first_name = "John"
last_name = "Smith"

full_name = first_name + " " + last_name
print(full_name)
````

````python
# More concatenation examples
department = "BI"
course_code = "303"

print(department + course_code)
print(department + " " + course_code)
print("Welcome to " + department + course_code + "!")
````

## 3. String Indexing

Each character in a string has an index number.

Important notes:
- Python starts counting from `0`.
- Negative indexes count from the end of the string.

Example: `Python`

| Character | P | y | t | h | o | n |
|----------|---|---|---|---|---|---|
| Index    | 0 | 1 | 2 | 3 | 4 | 5 |
| Negative | -6 | -5 | -4 | -3 | -2 | -1 |

````python
word = "Python"

print(word[0])   # First character
print(word[1])   # Second character
print(word[-1])  # Last character
print(word[-2])  # Second last character
````

````python
# More indexing examples
student_id = "BI303-2026"

print(student_id[0])
print(student_id[2])
print(student_id[-4])
print(student_id[-1])
````

## 4. String Slicing

Slicing is used to get part of a string.

Syntax:

```python
string[start:stop]
```

The `start` index is included, but the `stop` index is not included.

````python
text = "Data Science"

print(text[0:4])   # Characters from index 0 to 3
print(text[5:12])  # Characters from index 5 to 11
print(text[:4])    # From the beginning to index 3
print(text[5:])    # From index 5 to the end
````

````python
# More slicing examples
student_id = "BI303-2026"

print(student_id[0:2])   # BI
print(student_id[2:5])   # 303
print(student_id[6:])    # 2026
print(student_id[::-1])  # Reverse the string
````

## 5. Useful String Methods

A **method** is an action that belongs to an object. String methods help us clean, search, and transform text.

| Method | Description |
|--------|-------------|
| `.upper()` | Convert text to uppercase |
| `.lower()` | Convert text to lowercase |
| `.strip()` | Remove spaces from the beginning and end |
| `.replace(old, new)` | Replace part of the text |
| `.count(value)` | Count how many times a value appears |
| `.find(value)` | Find the index of a value |

````python
message = "  Python is fun. Python is useful.  "

print(message.upper())
print(message.lower())
print(message.strip())
print(message.replace("Python", "Programming"))
print(message.count("Python"))
print(message.find("fun"))
````

````python
# More string method examples
email = "  student@example.com  "
clean_email = email.strip().lower()

print(clean_email)
print(clean_email.startswith("student"))
print(clean_email.endswith(".com"))
print(clean_email.replace("example", "tbs"))
````

## 6. f-Strings

An **f-string** is a simple way to put variables inside a string. Put `f` before the opening quote and place variables inside `{}`.

````python
name = "Maya"
score = 87

print(f"Student name: {name}")
print(f"Score: {score}")
print(f"{name} got {score} points in the exam.")
````

````python
# More f-string examples
product = "Notebook"
price = 49.9
quantity = 3

print(f"Product: {product}")
print(f"Total price: {price * quantity}")
print(f"Total price: {price * quantity:.2f} baht")
````

## 7. Converting Values to Strings with `str()`

The `str()` function converts a value into a string.

This is useful when you want to combine text with numbers or other data types using `+`.

````python
age = 20
score = 95.5

print(str(age))
print(type(str(age)))

message = "Age: " + str(age) + ", Score: " + str(score)
print(message)
````

Without `str()`, Python will show an error when you try to join a string and a number using `+`.

````python
# This works because age is converted to a string first
age = 20
print("I am " + str(age) + " years old.")
````

````python
# More str() examples
order_id = 105
is_paid = True

print("Order ID: " + str(order_id))
print("Payment status: " + str(is_paid))
print("Data type after conversion:", type(str(is_paid)))
````

## 8. What Is a List?

A **list** is used to store multiple values in one variable. Lists are written with square brackets `[]`.

A list can contain strings, numbers, booleans, or even mixed data types.

````python
# Creating lists
fruits = ["apple", "banana", "orange"]
scores = [80, 75, 92, 88]
mixed_data = ["Alice", 20, True]

print(fruits)
print(scores)
print(mixed_data)
print(type(fruits))
````

````python
# More list examples
subjects = ["Python", "SQL", "Dashboard"]
prices = [120.5, 99.0, 250.75]
flags = [True, False, True]

print(subjects)
print(prices)
print(flags)
````

## 9. List Indexing

List indexing works like string indexing. The first item is at index `0`.

````python
fruits = ["apple", "banana", "orange", "mango"]

print(fruits[0])   # First item
print(fruits[2])   # Third item
print(fruits[-1])  # Last item
````

````python
# More list indexing examples
subjects = ["Python", "SQL", "Dashboard", "Excel"]

print(subjects[0])
print(subjects[3])
print(subjects[-1])
print(subjects[-3])
````

## 10. List Slicing

You can use slicing to get part of a list.

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[0:3])  # First three items
print(numbers[2:5])  # Items from index 2 to 4
print(numbers[:4])   # From the beginning to index 3
print(numbers[3:])   # From index 3 to the end
````

````python
# More list slicing examples
subjects = ["Python", "SQL", "Dashboard", "Excel", "Statistics"]

print(subjects[1:4])
print(subjects[:2])
print(subjects[2:])
print(subjects[::-1])
````

## 11. Updating List Items

Lists are **mutable**, which means you can change their values after creating them.

````python
fruits = ["apple", "banana", "orange"]

fruits[1] = "grape"

print(fruits)
````

````python
# More updating examples
subjects = ["Python", "SQL", "Dashboard"]

subjects[2] = "Power BI"
print(subjects)

subjects[0:2] = ["Basic Python", "Database"]
print(subjects)
````

## 12. Adding Items to a List

Common methods for adding items:

| Method | Description |
|--------|-------------|
| `.append(value)` | Add one item to the end |
| `.insert(index, value)` | Add one item at a specific position |
| `.extend(list)` | Add multiple items from another list |

````python
fruits = ["apple", "banana"]

fruits.append("orange")
print(fruits)

fruits.insert(1, "mango")
print(fruits)

fruits.extend(["grape", "melon"])
print(fruits)
````

````python
# More adding examples
tasks = ["read", "practice"]

tasks.append("submit")
tasks.insert(0, "login")
tasks.extend(["review", "reflect"])

print(tasks)
````

## 13. Removing Items from a List

Common methods for removing items:

| Method | Description |
|--------|-------------|
| `.remove(value)` | Remove a specific value |
| `.pop(index)` | Remove an item by index and return it |
| `.clear()` | Remove all items |

````python
fruits = ["apple", "banana", "orange", "mango"]

fruits.remove("banana")
print(fruits)

removed_item = fruits.pop(1)
print(f"Removed item: {removed_item}")
print(fruits)

fruits.clear()
print(fruits)
````

````python
# More removing examples
tasks = ["login", "read", "practice", "submit", "review"]

tasks.remove("read")
print(tasks)

last_task = tasks.pop()
print(last_task)
print(tasks)

del tasks[0]
print(tasks)
````

## 14. Useful List Functions and Methods

| Function or Method | Description |
|--------------------|-------------|
| `len(list)` | Count the number of items |
| `sum(list)` | Add all numeric items |
| `min(list)` | Get the smallest value |
| `max(list)` | Get the largest value |
| `.sort()` | Sort the list in place |
| `.reverse()` | Reverse the list in place |

````python
scores = [80, 75, 92, 88, 60]

print(len(scores))
print(sum(scores))
print(min(scores))
print(max(scores))
print(sum(scores) / len(scores))

scores.sort()
print(scores)

scores.reverse()
print(scores)
````

````python
# More list function examples
values = [12, 5, 30, 18, 7]

print("Sorted copy:", sorted(values))
print("Original list:", values)
print("Largest three values:", sorted(values, reverse=True)[:3])
print("Count of 5:", values.count(5))
````

## 15. 2D Lists

A **2D list** is a list inside another list. It is useful for storing table-like data, such as rows and columns.

Example structure:

```python
table = [
    [row_1_column_1, row_1_column_2],
    [row_2_column_1, row_2_column_2]
]
```

````python
# Creating a 2D list
students = [
    ["Alice", 85],
    ["Bob", 72],
    ["Charlie", 90]
]

print(students)
print(students[0])     # First row
print(students[0][0])  # First row, first column
print(students[0][1])  # First row, second column
````

### Accessing Rows and Columns

Use two indexes to access a value inside a 2D list:

```python
list_name[row_index][column_index]
```

The first index selects the row. The second index selects the item inside that row.

````python
# Accessing values in a 2D list
sales = [
    ["Jan", 12000, 9000],
    ["Feb", 15000, 11000],
    ["Mar", 18000, 14000]
]

print(sales[0][0])  # Month in row 0
print(sales[1][1])  # Revenue in row 1
print(sales[2][2])  # Cost in row 2
````

### Updating Values in a 2D List

Because lists are mutable, you can update values inside a 2D list by using row and column indexes.

````python
# Updating a value in a 2D list
sales = [
    ["Jan", 12000, 9000],
    ["Feb", 15000, 11000],
    ["Mar", 18000, 14000]
]

sales[1][1] = 16000

print(sales)
print(sales[1])
````

### 2D List Example Without Loops

You can calculate values from a small 2D list by accessing each row directly.

````python
# Calculate total revenue without using a loop
sales = [
    ["Jan", 12000, 9000],
    ["Feb", 15000, 11000],
    ["Mar", 18000, 14000]
]

total_revenue = sales[0][1] + sales[1][1] + sales[2][1]
total_cost = sales[0][2] + sales[1][2] + sales[2][2]
profit = total_revenue - total_cost

print(f"Total revenue: {total_revenue}")
print(f"Total cost: {total_cost}")
print(f"Profit: {profit}")
````

## 16. Splitting and Joining Strings

`split()` converts a string into a list.

`join()` combines list items into one string.

````python
sentence = "Python is easy to learn"
words = sentence.split()

print(words)

new_sentence = "-".join(words)
print(new_sentence)
````

### More `split()` Examples

The `split()` function can separate text by spaces, commas, symbols, or a limited number of breaks.

````python
# split() with spaces
sentence = "Python is easy to learn"
print(sentence.split())

# split() with comma
product_text = "apple,banana,orange"
print(product_text.split(","))

# split() with a custom symbol
student_record = "Alice|BI303|90"
print(student_record.split("|"))
````

````python
# split() with maxsplit
full_name = "John Michael Smith"
print(full_name.split(" ", 1))   # Split only once
print(full_name.split(" ", 2))   # Split two times

# split() together with strip()
raw_tags = " python, sql, dashboard "
tags = raw_tags.strip().split(", ")
print(tags)
````
