## 1. Why Do We Need Loops?

A **loop** lets Python repeat code many times.

Without a loop, repeated work becomes long and difficult to maintain.


````python
# Without a loop
print("Process customer 1")
print("Process customer 2")
print("Process customer 3")
print("Process customer 4")
print("Process customer 5")
````

````python
# With a loop
for customer_no in range(1, 6):
    print("Process customer", customer_no)
````

## 2. `for` Loop

Use a `for` loop when you know the collection you want to go through.

Basic syntax:

```python
for item in collection:
    # code to repeat
```

The indented code runs once for each item.


````python
products = ["Coffee", "Tea", "Juice", "Water"]

for product in products:
    print(product)
````

````python
sales = [1200, 950, 1450, 2100]

for amount in sales:
    print("Sale amount:", amount)
````

## 3. `range()`

`range()` creates a sequence of numbers. It is commonly used with `for` loops.

Useful patterns:

| Pattern | Meaning |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 |


````python
for number in range(5):
    print(number)
````

````python
for day in range(1, 8):
    print("Day", day)
````

````python
# Start, stop, step
for even_number in range(2, 11, 2):
    print(even_number)
````

````python
# Count down
for number in range(5, 0, -1):
    print(number)

print("Start!")
````

### Useful Command: Convert `range()` to a List

`range()` does not display all values by itself in many situations. Use `list()` when you want to see the generated numbers.


````python
print(range(5))
print(list(range(5)))
print(list(range(1, 11)))
print(list(range(10, 0, -2)))
````

## 4. Loop with Calculation

A common loop pattern is to create a variable before the loop, then update it inside the loop.

This is called an **accumulator**.


````python
sales = [1200, 950, 1450, 2100]
total_sales = 0

for amount in sales:
    total_sales = total_sales + amount

print("Total sales:", total_sales)
````

````python
sales = [1200, 950, 1450, 2100]
total_sales = 0

for amount in sales:
    total_sales += amount

average_sales = total_sales / len(sales)
print("Total sales:", total_sales)
print("Average sales:", average_sales)
````

### Useful Commands for Lists of Numbers

Python already has helpful functions for common calculations:

- `sum()` calculates total
- `len()` counts items
- `min()` finds the smallest value
- `max()` finds the largest value


````python
sales = [1200, 950, 1450, 2100]

print("Total:", sum(sales))
print("Count:", len(sales))
print("Lowest:", min(sales))
print("Highest:", max(sales))
print("Average:", sum(sales) / len(sales))
````

## 5. `enumerate()`

Use `enumerate()` when you need both:

- the index number
- the item value


````python
products = ["Coffee", "Tea", "Juice", "Water"]

for index, product in enumerate(products):
    print(index, product)
````

````python
products = ["Coffee", "Tea", "Juice", "Water"]

for position, product in enumerate(products, start=1):
    print(position, product)
````

## 6. String with Loop

A string is a sequence of characters, so you can loop through it one character at a time.


````python
word = "Python"

for character in word:
    print(character)
````

````python
course_code = "BI303"

for character in course_code:
    print(character, "is digit?", character.isdigit())
````

````python
text = "Business Intelligence"
vowel_count = 0

for character in text.lower():
    if character in "aeiou":
        vowel_count += 1

print("Text:", text)
print("Number of vowels:", vowel_count)
````

## 7. Loop with Condition

You can combine a loop with `if`, `elif`, and `else` to make decisions during repetition.


````python
sales = [1200, 950, 1450, 2100, 700]

for amount in sales:
    if amount >= 1500:
        print(amount, "High")
    elif amount >= 1000:
        print(amount, "Medium")
    else:
        print(amount, "Low")
````

````python
scores = [88, 42, 75, 91, 60, 35]
passed = 0
failed = 0

for score in scores:
    if score >= 50:
        passed += 1
    else:
        failed += 1

print("Passed:", passed)
print("Failed:", failed)
````

### Useful Commands: `continue` and `break`

- `continue` skips the rest of the current loop round and moves to the next item.
- `break` stops the loop immediately.


````python
# continue example: skip missing values
sales = [1200, None, 950, 1450, None, 2100]

total = 0
for amount in sales:
    if amount is None:
        continue
    total += amount

print("Total sales without missing values:", total)
````

````python
# break example: stop when the target is found
customers = ["Anna", "Ben", "Chai", "Dao", "Emma"]
target = "Chai"

for customer in customers:
    print("Checking", customer)
    if customer == target:
        print("Found", target)
        break
````

## 8. `while` Loop

Use a `while` loop when you want to repeat code as long as a condition is true.

Basic syntax:

```python
while condition:
    # code to repeat
```

Important: make sure something inside the loop changes the condition, or the loop may run forever.


````python
count = 1

while count <= 5:
    print("Count:", count)
    count += 1
````

````python
balance = 1000
withdrawal = 250
round_no = 1

while balance >= withdrawal:
    balance -= withdrawal
    print("Round", round_no, "remaining balance:", balance)
    round_no += 1

print("Stop because balance is lower than withdrawal amount")
````

### `while` Loop with a Flag Variable

A **flag** is a Boolean variable used to control whether a loop should continue.


````python
attempts = 0
logged_in = False
correct_pin = "1234"
entered_pins = ["1111", "2222", "1234"]

while not logged_in and attempts < len(entered_pins):
    pin = entered_pins[attempts]
    print("Try PIN:", pin)

    if pin == correct_pin:
        logged_in = True
        print("Login successful")
    else:
        print("Wrong PIN")

    attempts += 1
````

## 9. Function with Loop

Functions and loops work well together. Put repeated logic inside a function when you want to reuse it.


````python
def calculate_total(numbers):
    total = 0
    for number in numbers:
        total += number
    return total

sales_monday = [100, 200, 150]
sales_tuesday = [300, 250, 400]

print(calculate_total(sales_monday))
print(calculate_total(sales_tuesday))
````

````python
def count_passed(scores, pass_score=50):
    count = 0
    for score in scores:
        if score >= pass_score:
            count += 1
    return count

exam_scores = [88, 42, 75, 91, 60, 35]
print("Passed students:", count_passed(exam_scores))
print("Passed with score >= 80:", count_passed(exam_scores, 80))
````

````python
def clean_product_codes(codes):
    cleaned_codes = []
    for code in codes:
        cleaned_codes.append(code.strip().upper())
    return cleaned_codes

raw_codes = [" bi303 ", " py101", "data20 "]
print(clean_product_codes(raw_codes))
````

## 10. Loop in Loop: Nested Loops

A **nested loop** is a loop inside another loop.

Use nested loops when you need to combine two dimensions, such as products by days, rows by columns, or stores by months.


````python
products = ["Coffee", "Tea", "Juice"]
days = ["Mon", "Tue", "Wed"]

for product in products:
    for day in days:
        print(product, day)
````

````python
# Multiplication table
for row in range(1, 6):
    for col in range(1, 6):
        print(row * col, end="\t")
    print()
````

````python
# Sales table: stores x days
stores = ["Store A", "Store B"]
days = ["Mon", "Tue", "Wed"]
sales = [
    [1200, 950, 1100],
    [1500, 1700, 1600]
]

for store_index, store in enumerate(stores):
    print(store)
    for day_index, day in enumerate(days):
        print(" ", day, sales[store_index][day_index])
````

## 11. Useful Command: `zip()`

Use `zip()` when you want to loop through two or more lists at the same time.


````python
products = ["Coffee", "Tea", "Juice"]
prices = [60, 45, 55]

for product, price in zip(products, prices):
    print(product, "costs", price, "baht")
````

````python
products = ["Coffee", "Tea", "Juice"]
prices = [60, 45, 55]
quantities = [10, 8, 12]

for product, price, quantity in zip(products, prices, quantities):
    revenue = price * quantity
    print(product, "revenue:", revenue)
````

## 13. Lab 1: Daily Sales Summary

Complete the tasks below using loops.

Data:

```python
daily_sales = [1200, 950, 1450, 2100, 700, 1800, 1300]
```

Tasks:

1. Print each day and sale amount.
2. Calculate total sales.
3. Calculate average sales.
4. Count how many days have sales at least 1,000.
5. Find the highest sale amount without using `max()`.


````python
# Lab 1 starter code
daily_sales = [1200, 950, 1450, 2100, 700, 1800, 1300]

# 1. Print each day and sale amount
for day, amount in enumerate(daily_sales, start=1):
    print("Day", day, "sales:", amount)

# 2. Calculate total sales
total_sales = 0
for amount in daily_sales:
    total_sales += amount

print("Total sales:", total_sales)

# 3. Calculate average sales
average_sales = total_sales / len(daily_sales)
print("Average sales:", average_sales)

# 4. Count days with sales at least 1,000
high_day_count = 0
for amount in daily_sales:
    if amount >= 1000:
        high_day_count += 1

print("Days with sales >= 1000:", high_day_count)

# 5. Find highest sale amount without max()
highest_sale = daily_sales[0]
for amount in daily_sales:
    if amount > highest_sale:
        highest_sale = amount

print("Highest sale:", highest_sale)
````

## 15. Lab 3: Password Check with `while`

Use a `while` loop to simulate password attempts.

Rules:

- Correct password is `python303`.
- User has at most 3 attempts.
- Use the prepared list `attempted_passwords` instead of `input()` so the notebook runs automatically.


````python
# Lab 3 starter code
correct_password = "python303"
attempted_passwords = ["abc", "python", "python303"]
max_attempts = 3
attempt = 0
success = False

while attempt < max_attempts and not success:
    password = attempted_passwords[attempt]
    print("Attempt", attempt + 1, "password:", password)

    if password == correct_password:
        success = True
        print("Access granted")
    else:
        print("Access denied")

    attempt += 1

if not success:
    print("Account locked")
````
