## 1. What Is a Dictionary?

A **dictionary** stores data as **key-value pairs**.

Example:

```python
student = {
    "name": "Anna",
    "age": 20,
    "major": "Business"
}
```

- `"name"` is a key.
- `"Anna"` is a value.
- We use a key to find its value.

Dictionaries are useful when each piece of data has a label.

````python
student = {
    "name": "Anna",
    "age": 20,
    "major": "Business"
}

print(student)
print(type(student))
````

## 2. Create a Dictionary

A dictionary uses curly braces `{}`.

Each item follows this pattern:

```python
key: value
```

Keys are often strings, but values can be many data types: string, integer, float, boolean, list, or even another dictionary.

````python
product = {
    "name": "Wireless Mouse",
    "price": 450,
    "in_stock": True,
    "colors": ["black", "white", "blue"]
}

print(product)
````

## 3. Access Values

Use square brackets with the key name.

````python
student = {
    "name": "Anna",
    "age": 20,
    "major": "Business"
}

print(student["name"])
print(student["age"])
````

### Use `.get()` Safely

If the key does not exist, square brackets cause an error. The `.get()` method returns `None` or a default value instead.

````python
student = {
    "name": "Anna",
    "age": 20,
    "major": "Business"
}

print(student.get("name"))
print(student.get("email"))
print(student.get("email", "No email yet"))
````

## 4. Add and Update Values

Use the key and assignment operator `=`.

- If the key does not exist, Python adds it.
- If the key already exists, Python updates its value.

````python
student = {
    "name": "Anna",
    "age": 20,
    "major": "Business"
}

student["email"] = "anna@example.com"
student["age"] = 21

print(student)
````

## 5. Delete Values

Common ways to remove dictionary items:

| Code | Meaning |
|---|---|
| `del dictionary[key]` | Delete by key |
| `dictionary.pop(key)` | Delete by key and return the removed value |
| `dictionary.clear()` | Remove all items |

````python
student = {
    "name": "Anna",
    "age": 20,
    "major": "Business",
    "email": "anna@example.com"
}

del student["email"]
removed_age = student.pop("age")

print(student)
print("Removed age:", removed_age)
````

## 6. Check If a Key Exists

````python
student = {
    "name": "Anna",
    "age": 20,
    "major": "Business"
}

print("name" in student)
print("email" in student)

if "major" in student:
    print("Major:", student["major"])
````

## 9. Dictionary With Strings

Dictionaries are very useful for counting text.

Example: count how many times each character appears in a word.

````python
word = "banana"
letter_count = {}

for letter in word:
    if letter in letter_count:
        letter_count[letter] = letter_count[letter] + 1
    else:
        letter_count[letter] = 1

print(letter_count)
````

### Cleaner Version With `.get()`

`dict.get(key, default)` helps make counting code shorter.

````python
word = "banana"
letter_count = {}

for letter in word:
    letter_count[letter] = letter_count.get(letter, 0) + 1

print(letter_count)
````

### Count Words in a Sentence

We can combine string methods, loops, and dictionaries.

````python
sentence = "python is fun and python is useful"
words = sentence.split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)
````

## 11. Dictionary With Functions

Functions help us reuse dictionary logic.

````python
def describe_product(product):
    return product["name"] + " costs " + str(product["price"]) + " baht"

product = {
    "name": "Notebook",
    "price": 35
}

print(describe_product(product))
````

````python
def calculate_cart_total(cart):
    total = 0
    for item, price in cart.items():
        total = total + price
    return total

cart = {
    "coffee": 60,
    "cake": 120,
    "water": 10
}

print(calculate_cart_total(cart))
````

### Function That Updates a Dictionary

## 13. Dictionary Comprehension

A **dictionary comprehension** is a shorter way to create a dictionary from a loop.

Basic pattern:

```python
new_dict = {key_expression: value_expression for item in items}
```

````python
numbers = [1, 2, 3, 4, 5]
squares = {number: number ** 2 for number in numbers}

print(squares)
````

````python
products = ["coffee", "tea", "cake"]
name_lengths = {product: len(product) for product in products}

print(name_lengths)
````

### Dictionary Comprehension With Condition

````python
prices = {
    "coffee": 60,
    "tea": 45,
    "cake": 120,
    "water": 10
}

expensive = {item: price for item, price in prices.items() if price >= 60}

print(expensive)
````

## 15. Practical Use Case: Mini Store Inventory

This use case combines:

- dictionary
- dictionary inside dictionary
- list
- string methods
- functions
- loops

Goal: build a small inventory system.

````python
inventory = {
    "P001": {
        "name": "coffee",
        "category": "drink",
        "price": 60,
        "stock": 30
    },
    "P002": {
        "name": "tea",
        "category": "drink",
        "price": 45,
        "stock": 20
    },
    "P003": {
        "name": "cake",
        "category": "dessert",
        "price": 120,
        "stock": 10
    }
}

print(inventory)
````

### Function 1: Print Inventory

````python
def print_inventory(inventory):
    for product_id, product in inventory.items():
        name = product["name"].title()
        price = product["price"]
        stock = product["stock"]
        print(product_id, "-", name, "|", price, "baht", "| stock:", stock)

print_inventory(inventory)
````

### Function 2: Search Product by Name

````python
def search_product(inventory, keyword):
    keyword = keyword.lower().strip()
    results = []

    for product_id, product in inventory.items():
        product_name = product["name"].lower()
        if keyword in product_name:
            results.append({
                "product_id": product_id,
                "name": product["name"],
                "price": product["price"],
                "stock": product["stock"]
            })

    return results

print(search_product(inventory, " COF "))
````

### Function 3: Sell a Product

The function below checks stock, updates stock, and returns a receipt dictionary.

````python
def sell_product(inventory, product_id, quantity):
    if product_id not in inventory:
        return {"success": False, "message": "Product not found"}

    product = inventory[product_id]

    if quantity > product["stock"]:
        return {"success": False, "message": "Not enough stock"}

    product["stock"] = product["stock"] - quantity
    total = product["price"] * quantity

    return {
        "success": True,
        "product_name": product["name"].title(),
        "quantity": quantity,
        "total": total,
        "remaining_stock": product["stock"]
    }

receipt = sell_product(inventory, "P001", 3)
print(receipt)
print_inventory(inventory)
````

### Function 4: Summarize Inventory Value

Inventory value means:

```python
price * stock
```

for each product.

````python
def calculate_inventory_value(inventory):
    total_value = 0

    for product in inventory.values():
        total_value = total_value + (product["price"] * product["stock"])

    return total_value

print("Inventory value:", calculate_inventory_value(inventory), "baht")
````
