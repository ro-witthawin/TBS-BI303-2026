## 1. Why Do We Need Classes?

A **class** lets us combine related data and behavior in one place.

Example idea:


## 2. Class vs Object

A **class** is a blueprint.

An **object** is a real thing created from that blueprint.

| Word | Meaning | Example |
|---|---|---|
| Class | Blueprint/template | `Student` |
| Object | One real item from the class | `anna` |
| Attribute | Data inside an object | `name`, `major`, `gpa` |
| Method | Function inside a class | `show_profile()` |

````python
class Student:
    pass

anna = Student()

print(anna)
print(type(anna))
````

## 3. Add Attributes to an Object

An **attribute** is a variable that belongs to an object.

We can add attributes after creating an object, but later we will learn a better way using `__init__()`.

````python
class Student:
    pass

anna = Student()
anna.name = "Anna"
anna.major = "Business Intelligence"
anna.gpa = 3.50

print(anna.name)
print(anna.major)
print(anna.gpa)
````

## 4. Add a Method

A **method** is a function that belongs to a class.

The first parameter of an object method is usually named `self`.

`self` means: the current object that is using this method.

````python
class Student:
    def show_profile(self):
        print("Name:", self.name)
        print("Major:", self.major)
        print("GPA:", self.gpa)

anna = Student()
anna.name = "Anna"
anna.major = "Business Intelligence"
anna.gpa = 3.50

anna.show_profile()
````

## 5. The `__init__()` Method

`__init__()` is a special method that runs automatically when we create a new object.

It is commonly used to set starting attribute values.

````python
class Student:
    def __init__(self, name, major, gpa):
        self.name = name
        self.major = major
        self.gpa = gpa

    def show_profile(self):
        print("Name:", self.name)
        print("Major:", self.major)
        print("GPA:", self.gpa)

anna = Student("Anna", "Business Intelligence", 3.50)
anna.show_profile()
````

## 6. Create Multiple Objects

One class can create many objects.

Each object has its own attribute values.

````python
class Student:
    def __init__(self, name, major, gpa):
        self.name = name
        self.major = major
        self.gpa = gpa

    def show_profile(self):
        print(self.name, "studies", self.major, "with GPA", self.gpa)

anna = Student("Anna", "Business Intelligence", 3.50)
ben = Student("Ben", "Marketing", 3.20)
cara = Student("Cara", "Finance", 3.80)

anna.show_profile()
ben.show_profile()
cara.show_profile()
````

## 7. Methods Can Return Values

A method can use `return`, just like a normal function.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return self.price * self.quantity

keyboard = Product("Keyboard", 850, 12)
mouse = Product("Mouse", 450, 25)

print(keyboard.name, keyboard.total_value())
print(mouse.name, mouse.total_value())
````

## 8. Methods Can Receive Parameters

A method can receive extra information through parameters.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def sell(self, amount):
        self.quantity = self.quantity - amount

    def restock(self, amount):
        self.quantity = self.quantity + amount

    def show_stock(self):
        print(self.name, "stock:", self.quantity)

notebook = Product("Notebook", 35, 100)
notebook.show_stock()

notebook.sell(8)
notebook.show_stock()

notebook.restock(20)
notebook.show_stock()
````

## 10. Add Validation Logic

A method can protect an object from invalid changes.

In this example, we do not allow product stock to become negative.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def sell(self, amount):
        if amount <= self.quantity:
            self.quantity = self.quantity - amount
            print("Sold", amount, self.name)
        else:
            print("Not enough stock")

    def show_stock(self):
        print(self.name, "stock:", self.quantity)

tea = Product("Green Tea", 60, 10)
tea.sell(3)
tea.show_stock()

tea.sell(20)
tea.show_stock()
````

## 12. Example: Bank Account Class

A class is useful when data and actions belong together.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance = self.balance + amount
            print("Deposit successful")
        else:
            print("Amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Amount must be positive")
        elif amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance = self.balance - amount
            print("Withdraw successful")

    def show_balance(self):
        print(self.owner, "balance:", self.balance)

account = BankAccount("Anna", 1000)
account.show_balance()
account.deposit(500)
account.withdraw(300)
account.show_balance()
````

## 13. Lists of Objects

Objects can be stored in lists, just like strings, numbers, and dictionaries.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return self.price * self.quantity

products = [
    Product("Coffee", 120, 10),
    Product("Tea", 80, 15),
    Product("Juice", 65, 20),
]

for product in products:
    print(product.name, product.total_value())
````

````python
total_inventory_value = 0

for product in products:
    total_inventory_value = total_inventory_value + product.total_value()

print("Total inventory value:", total_inventory_value)
````

## 14. Convert Objects to Dictionaries

Sometimes we need to convert an object to a dictionary before saving or exporting data.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def to_dict(self):
        return {
            "name": self.name,
            "price": self.price,
            "quantity": self.quantity,
            "total_value": self.price * self.quantity
        }

product = Product("Coffee", 120, 10)
print(product.to_dict())
````

## 15. Class Variables

An **instance variable** belongs to one object.

A **class variable** belongs to the class and is shared by all objects.

````python
class Employee:
    company_name = "TBS Company"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_info(self):
        print(self.name, self.salary, self.company_name)

employee1 = Employee("Anna", 30000)
employee2 = Employee("Ben", 35000)

employee1.show_info()
employee2.show_info()

Employee.company_name = "TBS International"

employee1.show_info()
employee2.show_info()
````

## 17. Encapsulation Basics

**Encapsulation** means keeping data and the methods that work with that data together.

Python also uses naming conventions:

| Name style | Meaning |
|---|---|
| `balance` | normal public attribute |
| `_balance` | internal-use attribute by convention |

The underscore does not make the value fully private, but it tells other programmers to use methods instead of changing it directly.

````python
class Wallet:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        if amount > 0:
            self._balance = self._balance + amount

    def get_balance(self):
        return self._balance

wallet = Wallet("Anna", 500)
wallet.deposit(250)

print(wallet.owner)
print(wallet.get_balance())
````

## 18. Inheritance Basics

**Inheritance** lets one class reuse another class.

The child class receives attributes and methods from the parent class.

````python
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email

    def show_profile(self):
        print(self.username, self.email)

class AdminUser(User):
    def delete_user(self, username):
        print(self.username, "deleted user", username)

admin = AdminUser("admin01", "admin@example.com")
admin.show_profile()
admin.delete_user("old_user")
````

## 19. Override a Method

A child class can replace a parent method with its own version.

````python
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email

    def role(self):
        return "Normal user"

class AdminUser(User):
    def role(self):
        return "Admin user"

user = User("anna", "anna@example.com")
admin = AdminUser("admin01", "admin@example.com")

print(user.username, user.role())
print(admin.username, admin.role())
````

## 20. Mini Case Study: CRM Contact

A CRM system stores customer contacts. Each contact has data and useful actions.

````python
class Contact:
    def __init__(self, first_name, last_name, email, tags=None):
        self.first_name = first_name
        self.last_name = last_name
        self.email = email
        self.tags = tags or []

    def full_name(self):
        return self.first_name + " " + self.last_name

    def add_tag(self, tag):
        if tag not in self.tags:
            self.tags.append(tag)

    def show_contact(self):
        print("Name:", self.full_name())
        print("Email:", self.email)
        print("Tags:", self.tags)

contact = Contact("Anna", "Smith", "anna@example.com")
contact.add_tag("lead")
contact.add_tag("newsletter")
contact.show_contact()
````

## 21. Mini Case Study: Sales Report

Objects can make business calculations easier to organize.

````python
class Sale:
    def __init__(self, product_name, unit_price, quantity):
        self.product_name = product_name
        self.unit_price = unit_price
        self.quantity = quantity

    def revenue(self):
        return self.unit_price * self.quantity

sales = [
    Sale("Coffee", 120, 20),
    Sale("Tea", 80, 15),
    Sale("Juice", 65, 30),
]

for sale in sales:
    print(sale.product_name, sale.revenue())

print("Total revenue:", sum(sale.revenue() for sale in sales))
````

## 22. Common Mistakes

### Mistake 1: Forgetting `self`

```python
class Student:
    def show_name():
        print(self.name)
```

The method should receive `self` as the first parameter.

````python
class Student:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print(self.name)

student = Student("Anna")
student.show_name()
````

### Mistake 2: Using the Class Instead of an Object

The class is the blueprint. Create an object before using object attributes.

````python
class Product:
    def __init__(self, name):
        self.name = name

product = Product("Coffee")
print(product.name)
````

### Mistake 3: Forgetting Parentheses When Creating an Object

Use `Product("Coffee")`, not just `Product`.

````python
class Product:
    def __init__(self, name):
        self.name = name

product_class = Product
product_object = Product("Coffee")

print(product_class)
print(product_object.name)
````
